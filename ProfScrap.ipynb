Ahora todo esto lo quiero dentro de un CSV, por lo que todo esto lo meto dentro de la estructura de creación de CSV, además al final del for, incluir que todos los datos que se obtengan se escriban el csv.

```python
with open('UniScrap.csv', 'w', newline='') as csvfile:
    csvwriter = csv.writer(csvfile)
    csvwriter.writerow(['Nombre', 'Ciudad', 'Estado', 'Numero de Profesores', 'link']) # meto una lista por que writerow necesita un solo objeto, entonces tengo que entregarle varios objetos en 1

    page = 1
    while True:
        url = f'https://www.misprofesores.com/Universidades?pag={page}'
        response = requests.get(url)
        soup = BeautifulSoup(response.content, 'html.parser')

        datos_univ = soup.find_all('tr', {"bgcolor": True})
        # si quisiera buscar explicitamente un atributo es con
        # soup.find_all('etiqueta xd', {'atributo xd': 'valor del atributo xd'})

        if not datos_univ:
            break

        for universidad in datos_univ:
            #universidad_name= universidad.find('p', class_="indent").text
            #universidad_link= universidad.find('a', {"href": True})['href'] #ubico y extraigo link por attrs
            universidad_td = universidad.find_all('td') # hay 4 td en el html, extraigo datos en orden

            universidad_name = universidad_td[0].text
            universidad_link =universidad_td[0].find('a')['href'] #extraigo explicitamente el link
            universidad_city = universidad_td[1].text
            universidad_state = universidad_td[2].text
            profs = universidad_td[3].text
            csvwriter.writerow([universidad_name, universidad_city, universidad_state, profs, universidad_link]) #metotodo lo almacenado en las variables en la lista de writerow
        page = page + 1
```
